```python
"""
╔══════════════════════════════════════════════════════════════════════════╗
║        TESS – Emotion Classification with Bidirectional LSTM            ║
║        Kaggle GPU: T4 x2  |  TensorFlow MirroredStrategy               ║
╚══════════════════════════════════════════════════════════════════════════╝
Features : MFCC (40) + Δ-MFCC (40) + ΔΔ-MFCC (40) = 120 per frame
Model    : 3-layer BiLSTM → BatchNorm/Dropout → Dense head → Softmax(7)
Outputs  : confusion_matrix.png | acc_loss_curves.png | tess_lstm_best.h5
"""

# ─── IMPORTS ─────────────────────────────────────────────────────────────────
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import librosa
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Bidirectional, LSTM, Dense, Dropout, BatchNormalization
)
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import (
    ModelCheckpoint, EarlyStopping, ReduceLROnPlateau
)
from tensorflow.keras.optimizers import Adam

print(f"TensorFlow version : {tf.__version__}")
print(f"GPUs available     : {tf.config.list_physical_devices('GPU')}")

# ─── HARD-CODED CONFIG ───────────────────────────────────────────────────────
DATA_PATH    = (
    "/kaggle/input/datasets/ejlok1/"
    "toronto-emotional-speech-set-tess/"
    "TESS Toronto emotional speech set data"
)
SAMPLE_RATE  = 22050
DURATION     = 3          # seconds to load per clip
N_MFCC       = 40         # MFCC coefficients  → ×3 with deltas = 120 features
MAX_LEN      = 130        # time-steps (frames) after pad / truncate
BATCH_SIZE   = 256        # large batch for T4x2 (128 per GPU via MirroredStrategy)
EPOCHS       = 100
LR           = 1e-3
WORKING_DIR  = "/kaggle/working"
MODEL_PATH   = os.path.join(WORKING_DIR, "tess_lstm_best.h5")

# Canonical 7-class label set
EMOTION_LABELS = [
    "angry", "disgust", "fear",
    "happy", "neutral", "pleasant_surprise", "sad",
]

# ─── FOLDER → LABEL MAPPING ──────────────────────────────────────────────────
def folder_to_label(folder_name: str) -> str | None:
    """
    Converts TESS folder names (e.g. 'OAF_angry', 'YAF_pleasant_surprised')
    to one of the 7 canonical labels.
    """
    fn = folder_name.lower()
    if   "pleasant" in fn:  return "pleasant_surprise"
    elif "angry"    in fn:  return "angry"
    elif "disgust"  in fn:  return "disgust"
    elif "fear"     in fn:  return "fear"
    elif "happy"    in fn:  return "happy"
    elif "neutral"  in fn:  return "neutral"
    elif "sad"      in fn:  return "sad"
    return None

# ─── FEATURE EXTRACTION ──────────────────────────────────────────────────────
def extract_mfcc_sequence(file_path: str) -> np.ndarray | None:
    """
    Returns an array of shape (MAX_LEN, N_MFCC * 3):
      - 40 MFCCs
      - 40 Δ-MFCCs
      - 40 ΔΔ-MFCCs
    padded / truncated to MAX_LEN time-steps.
    """
    try:
        audio, sr = librosa.load(file_path, sr=SAMPLE_RATE, duration=DURATION)

        mfcc    = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=N_MFCC)  # (40, T)
        delta1  = librosa.feature.delta(mfcc)                           # (40, T)
        delta2  = librosa.feature.delta(mfcc, order=2)                  # (40, T)

        features = np.concatenate([mfcc, delta1, delta2], axis=0)       # (120, T)
        features = features.T                                            # (T, 120)

        # Pad or truncate to MAX_LEN
        if features.shape[0] < MAX_LEN:
            pad = MAX_LEN - features.shape[0]
            features = np.pad(features, ((0, pad), (0, 0)), mode="constant")
        else:
            features = features[:MAX_LEN]

        return features.astype(np.float32)

    except Exception as exc:
        print(f"  ⚠  Error [{file_path}]: {exc}")
        return None

# ─── LOAD DATASET ────────────────────────────────────────────────────────────
def load_tess(data_path: str):
    X_list, y_list = [], []
    folders = sorted(os.listdir(data_path))

    for folder in folders:
        folder_path = os.path.join(data_path, folder)
        if not os.path.isdir(folder_path):
            continue

        label = folder_to_label(folder)
        if label is None:
            print(f"  ⚠  Skipping unknown folder: {folder}")
            continue

        wav_files = [f for f in os.listdir(folder_path) if f.endswith(".wav")]
        print(f"  📂  {folder:<30}  →  '{label}'  ({len(wav_files)} files)")

        for wav in wav_files:
            seq = extract_mfcc_sequence(os.path.join(folder_path, wav))
            if seq is not None:
                X_list.append(seq)
                y_list.append(label)

    return np.array(X_list, dtype=np.float32), np.array(y_list)


print("\n" + "─" * 60)
print("  Loading TESS dataset & extracting features …")
print("─" * 60)
X, y_raw = load_tess(DATA_PATH)

print(f"\n  ✅  Total samples  : {len(X)}")
print(f"  ✅  Feature shape  : {X.shape}   →  (samples, {MAX_LEN} frames, {N_MFCC*3} features)")
print("\n  Emotion distribution:")
dist = pd.Series(y_raw).value_counts().sort_index()
print(dist.to_string())

# ─── ENCODE LABELS ───────────────────────────────────────────────────────────
le        = LabelEncoder()
y_int     = le.fit_transform(y_raw)                  # integer labels
y_cat     = to_categorical(y_int)                    # one-hot
NUM_CLASS = y_cat.shape[1]

print(f"\n  Classes ({NUM_CLASS}): {list(le.classes_)}")

# ─── TRAIN / VAL / TEST SPLIT  (70 / 10 / 20) ───────────────────────────────
X_trainval, X_test, y_trainval, y_test = train_test_split(
    X, y_cat, test_size=0.20, random_state=42, stratify=y_int
)
y_trainval_int = np.argmax(y_trainval, axis=1)
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval, y_trainval, test_size=0.125,   # 0.125 × 0.80 ≈ 0.10 overall
    random_state=42, stratify=y_trainval_int
)

print(f"\n  Split — Train: {len(X_train)} | Val: {len(X_val)} | Test: {len(X_test)}")

# ─── MODEL (MirroredStrategy across both T4 GPUs) ────────────────────────────
strategy = tf.distribute.MirroredStrategy()
print(f"\n  GPUs in use : {strategy.num_replicas_in_sync}")

INPUT_SHAPE = (MAX_LEN, N_MFCC * 3)   # (130, 120)

with strategy.scope():
    model = Sequential(
        [
            # ── Layer block 1 ──────────────────────────────────────────────
            Bidirectional(
                LSTM(256, return_sequences=True),
                input_shape=INPUT_SHAPE,
            ),
            BatchNormalization(),
            Dropout(0.30),

            # ── Layer block 2 ──────────────────────────────────────────────
            Bidirectional(LSTM(128, return_sequences=True)),
            BatchNormalization(),
            Dropout(0.30),

            # ── Layer block 3 ──────────────────────────────────────────────
            Bidirectional(LSTM(64, return_sequences=False)),
            BatchNormalization(),
            Dropout(0.30),

            # ── Dense head ─────────────────────────────────────────────────
            Dense(256, activation="relu"),
            BatchNormalization(),
            Dropout(0.40),

            Dense(128, activation="relu"),
            Dropout(0.30),

            Dense(NUM_CLASS, activation="softmax"),
        ],
        name="BiLSTM_Emotion_Classifier",
    )

    model.compile(
        optimizer=Adam(learning_rate=LR),
        loss="categorical_crossentropy",
        metrics=["accuracy"],
    )

model.summary()

# ─── CALLBACKS ───────────────────────────────────────────────────────────────
callbacks = [
    ModelCheckpoint(
        MODEL_PATH,
        monitor="val_accuracy",
        save_best_only=True,
        verbose=1,
    ),
    EarlyStopping(
        monitor="val_accuracy",
        patience=18,
        restore_best_weights=True,
        verbose=1,
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=8,
        min_lr=1e-6,
        verbose=1,
    ),
]

# ─── TRAINING ────────────────────────────────────────────────────────────────
print("\n" + "─" * 60)
print(f"  Training  |  batch={BATCH_SIZE}  epochs={EPOCHS}")
print("─" * 60)

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=callbacks,
    verbose=1,
)

# ─── TEST EVALUATION ─────────────────────────────────────────────────────────
test_loss, test_acc = model.evaluate(X_test, y_test, batch_size=BATCH_SIZE, verbose=0)

print("\n" + "=" * 60)
print(f"  ✅  Test Accuracy : {test_acc * 100:.2f} %")
print(f"  ✅  Test Loss     : {test_loss:.4f}")
print("=" * 60)

# ─── CLASSIFICATION REPORT ───────────────────────────────────────────────────
y_pred      = model.predict(X_test, batch_size=BATCH_SIZE, verbose=0)
y_pred_int  = np.argmax(y_pred, axis=1)
y_true_int  = np.argmax(y_test,  axis=1)

print("\n📊  Classification Report\n")
print(
    classification_report(
        y_true_int, y_pred_int,
        target_names=le.classes_,
        digits=4,
    )
)

# ─── CONFUSION MATRIX ────────────────────────────────────────────────────────
cm = confusion_matrix(y_true_int, y_pred_int)

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=le.classes_,
    yticklabels=le.classes_,
    linewidths=0.5,
    ax=ax,
)
ax.set_title(
    f"Confusion Matrix — TESS BiLSTM Emotion Classifier\n"
    f"Test Accuracy: {test_acc * 100:.2f}%",
    fontsize=13, fontweight="bold", pad=14,
)
ax.set_ylabel("True Label",      fontsize=11)
ax.set_xlabel("Predicted Label", fontsize=11)
plt.xticks(rotation=40, ha="right", fontsize=9)
plt.yticks(rotation=0,  fontsize=9)
plt.tight_layout()
cm_path = os.path.join(WORKING_DIR, "confusion_matrix.png")
plt.savefig(cm_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"  ✅  Confusion matrix saved → {cm_path}")

# ─── ACCURACY / LOSS CURVES ──────────────────────────────────────────────────
epochs_ran = range(1, len(history.history["accuracy"]) + 1)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle(
    "TESS BiLSTM — Training History",
    fontsize=14, fontweight="bold", y=1.02,
)

# Accuracy subplot
axes[0].plot(epochs_ran, history.history["accuracy"],
             label="Train", color="steelblue", linewidth=2)
axes[0].plot(epochs_ran, history.history["val_accuracy"],
             label="Validation", color="orangered", linewidth=2, linestyle="--")
axes[0].set_title("Accuracy", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Epoch");  axes[0].set_ylabel("Accuracy")
axes[0].legend(fontsize=10);  axes[0].grid(True, alpha=0.3)
axes[0].set_ylim([0, 1.05])

# Loss subplot
axes[1].plot(epochs_ran, history.history["loss"],
             label="Train", color="steelblue", linewidth=2)
axes[1].plot(epochs_ran, history.history["val_loss"],
             label="Validation", color="orangered", linewidth=2, linestyle="--")
axes[1].set_title("Loss", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Epoch");  axes[1].set_ylabel("Loss")
axes[1].legend(fontsize=10);  axes[1].grid(True, alpha=0.3)

plt.tight_layout()
curve_path = os.path.join(WORKING_DIR, "acc_loss_curves.png")
plt.savefig(curve_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"  ✅  Accuracy/Loss curves saved → {curve_path}")

# ─── SAVE MODEL & LABEL ENCODER ──────────────────────────────────────────────
model.save(MODEL_PATH)
np.save(os.path.join(WORKING_DIR, "label_classes.npy"), le.classes_)

print("\n" + "═" * 60)
print(f"  Model saved        → {MODEL_PATH}")
print(f"  Label classes      → {WORKING_DIR}/label_classes.npy")
print(f"  Confusion matrix   → {cm_path}")
print(f"  Acc/Loss curves    → {curve_path}")
print("═" * 60)
print("\n  🎉  Done!\n")
```
